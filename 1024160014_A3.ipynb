```python
#q6
data={
    "employee_id":[101,102,103,104,105],
    "name":["Alice",'Bob','Charlie','Diana','Edward'],
    'Department':["HR","IT","IT","Marketing","Sales"],
    "Age":[29,34,41,28,38],
    "Salary":[50000,70000,65000,55000,60000],
    "Year_of_Experience":[4,8,10,3,12],
    "joinig_date":["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-15"],
    "Gender":["female","male","male","female","male"],
    "bonus":[5000,7000,6000,4500,5000],
    "rating":[4.5,4.0,3.8,4.7,3.5]

}
df=pd.DataFrame(data)
print(df)
```

```python
df.shape
df.info()
df.describe()
df.head()
print("            \n       ")
df.tail(3)
```

```python
#e

print(df["Salary"].mean())
print(df["bonus"].sum())
print(df["Age"].min())
print(df['rating'].max())
```

```python
# f)
df = df.sort_values(by="Salary", ascending=False)
print(df)


# g)


df["rating"] = pd.to_numeric(df["rating"])

df["Performance_Category"] = df["rating"].apply(
    lambda x: "Excellent" if x >= 4.5
    else "Good" if x >= 4.0
    else "Average"
)

print(df)

# h)
print(df.isnull().sum())


# i)
df = df.rename(columns={"Employee_ID": "ID"})
print(df)


# j)
print(df[df["Year_of_Experience"] > 5])
print(df[df["Department"] == "IT"])


# k)
df["Tax"] = df["Salary"] * 0.10
print(df)


# l)
df.to_csv("modified_employee_data.csv", index=False)
print("File saved successfully.")
```
