```python
#q3

import random
#3.1
random.seed(1024160014)
numbers=[]
for i in range(100):
  numbers.append(random.randint(1,100))
print(numbers)

#3.2
odd= [x for x in numbers if x%2!=0]
print(len(odd))
print(odd)

#3.3
even=[x for x in numbers if x%2==0]
print(len(even))
print(even)

#3.4
def isPrime(n):
  if n <=1:
    return False
  for i in range(2,int(n**0.5)+1):
    if n%i==0:
      return False
  return True
prime=[x for x in numbers if isPrime(x)]
print(len(prime))
print(prime)
#3.5
from collections import Counter
freq = Counter(numbers)
freq.most_common(5)
```
